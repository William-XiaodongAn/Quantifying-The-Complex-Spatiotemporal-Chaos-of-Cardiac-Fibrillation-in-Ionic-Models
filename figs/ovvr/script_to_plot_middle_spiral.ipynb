```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os
# Replace with your own image paths
image_names = ["Kr_0_dot_3_spiral_400ms.png","Kr_0_dot_425_spiral_400ms.png","Kr_0_dot_5_spiral_400ms (2).png"]
labels = ["A", 'B', "C"]


fig, axes = plt.subplots(1, 3, figsize=(9, 3))

order = 1
for ax, path, label in zip(axes, image_names, labels):

    img = mpimg.imread(path)
    ax.imshow(img)
    ax.axis("off")

    # Add label at middle bottom
    ax.text(
        0.5, -0.05, label, color="red", fontsize=20, weight="bold",
        ha="center", va="top", transform=ax.transAxes
    )

    outname = 'ovvr_spiral.png'
    
plt.subplots_adjust(wspace=0.03, hspace=0)  # minimize spacing
plt.savefig(outname, dpi=300, bbox_inches="tight")
plt.show()
```
