```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os
# Replace with your own image paths
image_names = ["chaos_spiral_400ms.png", "quasichaos_spiral_400ms.png", "quasiperiod_spiral_400ms.png", "period_spiral_400ms.png"]
labels = ["A", 'B$_1$', 'B$_2$', "C"]


folder_path = "taud" 
order_lst = [1,-1,-1,1,-1]
folder = [ "taud" , "taur" , "tau0" , "tausi" , "TNNP"]
for idx,folder_path in enumerate(folder):
    fig, axes = plt.subplots(1, 4, figsize=(12, 3))
    
    order = order_lst[idx]
    image_paths = [os.path.join(folder_path, name) for name in image_names[::order]]
    for ax, path, label in zip(axes, image_paths, labels):
        img = mpimg.imread(path)
        ax.imshow(img)
        ax.axis("off")

        # Add label at middle bottom
        ax.text(
            0.5, -0.05, label, color="red", fontsize=20, weight="bold",
            ha="center", va="top", transform=ax.transAxes
        )

    outname = folder_path[:3] + '_' + folder_path[3:] + '_spiral.png'
    out_path = os.path.join(folder_path, outname)
    
    plt.subplots_adjust(wspace=0.03, hspace=0)  # minimize spacing
    plt.savefig(out_path, dpi=300, bbox_inches="tight")
    plt.show()
```
